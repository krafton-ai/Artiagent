```python
import os
from PIL import Image
import matplotlib.pyplot as plt

# List your 4 parent directories here
parent_dir = '/data3/jhpark/fusion/exp7'  # Change this to your actual parent directory

def list_subdirectories(parent_dir):
    return [
        name for name in os.listdir(parent_dir)
        if os.path.isdir(os.path.join(parent_dir, name)) and 'logs' not in name
    ]

flux_dirs = list_subdirectories(parent_dir)
flux_results_iter = iter(flux_dirs)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(15, 15))
fname = next(flux_results_iter)
generated_path = os.path.join(parent_dir, fname, 'generated.png')
img = Image.open(generated_path)
ax.imshow(img)
```
